```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("raphaelmanayon/temperature-and-ice-cream-sales")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os

# List files in the downloaded path to find the correct CSV file name
files_in_path = os.listdir(path)
csv_files = [f for f in files_in_path if f.endswith('.csv')]

if csv_files:
    file_name = csv_files[0] # Assuming there's only one CSV file or taking the first one
    print(f"Found CSV file: {file_name}")
else:
    raise FileNotFoundError("No CSV file found in the downloaded dataset directory.")

data_path = os.path.join(path, file_name)

# Load the dataset
df = pd.read_csv(data_path)

display(df.head())
```

Now, let's create a linear regression plot to visualize the relationship between `Temperature` and `Ice Cream Sales`.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a linear regression plot
plt.figure(figsize=(10, 6))
sns.regplot(x='Temperature', y='Ice Cream Profits', data=df, scatter_kws={'alpha':0.6}, line_kws={'color':'red'})
plt.title('Linear Regression Plot of Temperature vs. Ice Cream Profits')
plt.xlabel('Temperature')
plt.ylabel('Ice Cream Profits')
plt.grid(True)
plt.show()
```
